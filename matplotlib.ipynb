```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5,6,7]
sales = [1200, 1800, 1500, 2400, 3000,2500,2800]

plt.plot(days, sales, marker="s", linewidth=4)

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")
plt.grid(True)
plt.xticks(days)

plt.show()
```

```python
plt.figure(figsize=(6, 4))
colors = ["#eb34c0"] * len(sales)
colors[sales.index(max(sales))] = "orange"
bars = plt.bar(days, sales, width=0.3, edgecolor="black", color=colors)

plt.bar_label(bars,padding=5)
plt.title("Daily Sales",fontweight="bold",pad=3)
plt.xlabel("Day",fontsize=12)
plt.ylabel("Sales (₹)",fontsize=12)
plt.xticks(days)
plt.ylim(0, max(sales) * 1.15)
plt.grid(axis="y", linestyle="--", alpha=0.8)
plt.gca().set_axisbelow(True)
plt.tight_layout()
plt.show()
```

put two charts in one figure


```python
fig,ax = plt.subplots(1,2,figsize=(10,4))
# bar chart
bars=ax[0].bar(days, sales, width=0.3, edgecolor="black", color="#eb34c0")
ax[0].bar_label(bars,padding=5)
ax[0].set_title("Daily Sales",fontweight="bold",pad=3)
ax[0].set_xlabel("Day",fontsize=12)
ax[0].set_ylabel("Sales (₹)",fontsize=12)
ax[0].set_xticks(days)
ax[0].set_ylim(0, max(sales)*1.15)
ax[0].grid(axis="y",linestyle="--",alpha=0.8)
# plot chart
ax[1].plot(days, sales, marker="s", linewidth=4,color="#34c6eb")
ax[1].set_title("Daily Sales")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Sales (₹)")
ax[1].grid(True)
ax[1].set_xticks(days)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5, 6, 7]
sales = [1200, 1800, 1500, 2400, 3000, 2500, 2800]

fig,ax=plt.subplots(1,2,figsize=(10,4))
# bar chart
bars=ax[0].bar(days,sales,edgecolor="black",color="#eb343a")
ax[0].bar_label(bars,padding=5)
ax[0].set_title("Daily Sales",fontsize=20,fontweight="bold")
ax[0].set_xlabel("Day",fontsize=12)
ax[0].set_ylabel("Sales (₹)",fontsize=12)
ax[0].set_ylim(0,max(sales)*1.15)
ax[0].grid(axis="y")

# plots
ax[1].plot(days,sales,marker="s",linestyle="solid",color="#5534eb",linewidth=4)
ax[1].set_title("daily sales",fontsize=20,fontweight="bold")
ax[1].set_xlabel("day",fontsize=12)
ax[1].set_ylabel("sales(₹)",fontsize=12)
ax[1].grid(True)
plt.tight_layout()
plt.show()

```

histograms


```python
import matplotlib.pyplot as plt

sales = [1200, 1800, 1500, 2400, 3000, 2500, 2800]

plt.hist(sales, bins=6, edgecolor="black", color="#5534eb")
plt.title("Sales Distribution")
plt.xlabel("Sales (₹)")
plt.ylabel("Number of days")
plt.show()
```

scattered plots

```python
import matplotlib.pyplot as plt

ad_spend = [200, 400, 300, 600, 800, 500, 700]
sales = [1200, 1800, 1500, 2400, 3000, 2500, 2800]

plt.scatter(ad_spend, sales, color="#5534eb", s=150, alpha=0.4)
plt.title("Advertising Spend vs Sales")
plt.xlabel("Advertising Spend (₹)")
plt.ylabel("Sales (₹)")
plt.grid(True)
plt.show()
```

PIE CHARTS

```python
import matplotlib.pyplot as plt

products = ["Phones", "Laptops", "Accessories"]
sales = [5000, 3000, 2000]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0,0,0.15],
    colors=["#5534eb","#eb34cc","#34ebb7"],
    wedgeprops={"edgecolor":"black","linewidth":2})

plt.title("Sales by Product",fontsize=16,fontweight="bold")
plt.axis("equal")
plt.show()
```
